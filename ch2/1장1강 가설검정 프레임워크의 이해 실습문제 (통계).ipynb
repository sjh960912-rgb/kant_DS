## 필수 1. 평균 판매가격에 대한 양측검정

### 문제 1-1. 평균 판매가격이 180,000달러와 다른가?

#### 문제 설명

Ames 지역의 주택 평균 판매가격에 대한 기존 기준이 **180,000달러**라고 가정합니다.  
제공된 표본의 평균 판매가격이 이 기준과 다른지 단일표본 t검정으로 확인하세요.

#### 요구사항

1. `SalePrice`에서 결측치를 제거하여 `sale_price`에 저장하세요.
2. 표본 수와 표본평균을 출력하세요.
3. 다음 가설을 코드 주석이나 마크다운으로 작성하세요.
   - H₀: Ames 지역의 평균 판매가격은 180,000달러이다.
   - H₁: Ames 지역의 평균 판매가격은 180,000달러가 아니다.
4. 유의수준 `alpha`를 0.05로 설정하세요.
5. `stats.ttest_1samp()`와 `alternative="two-sided"`를 사용해 양측검정을 수행하세요.
6. t통계량과 p-value를 소수점 넷째 자리까지 출력하세요.
7. p-value와 유의수준을 비교해 귀무가설 기각 여부를 출력하세요.

#### 해석 질문

**Q1.** 이 문제에서 양측검정을 사용하는 이유는 무엇인가요?  
-> 방향이 정해진 주장이 아니기 때문에 양측검정을 사용
**Q2.** 검정 결과에 따라 귀무가설을 기각할 수 있나요?  
-> 네
**Q3.** 귀무가설을 기각하지 못했다면 평균 판매가격이 정확히 180,000달러라고 증명된 것인가요?
-> 충분한 증거를 찾지 못했다는 뜻이지 맞다는 걸 증명했다는 뜻은 아님

#### 제출 결과

- 가설 설정
- 표본 수와 표본평균
- 단일표본 t검정 코드
- t통계량과 p-value
- 귀무가설 판단
- Q1~Q3 답변

```python
# 필수 1 코드를 작성하세요.

import pandas as pd
from scipy import stats

# 1. SalePrice에서 결측치 제거하여 sale_price에 저장
sale_price = df['SalePrice'].dropna()

# 2. 표본 수와 표본평균 출력
n_sample = len(sale_price)
mean_sample = sale_price.mean()
print(f"표본 수: {n_sample}")
print(f"표본평균: {mean_sample:.4f}")

# 3. 가설 설정
# H0: Ames 지역의 평균 판매가격은 180,000달러이다.
# H1: Ames 지역의 평균 판매가격은 180,000달러가 아니다.

# 4. 유의수준 설정
alpha = 0.05

# 5. 단일표본 t검정 (양측검정)
t_stat, p_value = stats.ttest_1samp(sale_price, popmean=180000, alternative='two-sided')

# 6. t통계량과 p-value 출력 (소수점 넷째 자리까지)
print(f"t통계량: {t_stat:.4f}")
print(f"p-value: {p_value:.4f}")

# 7. 귀무가설 기각 여부 판단
if p_value < alpha:
    print(f"p-value({p_value:.4f}) < alpha({alpha}) → 귀무가설 기각")
else:
    print(f"p-value({p_value:.4f}) >= alpha({alpha}) → 귀무가설 기각 실패")
```

### 문제 2-1. 평균 품질 점수는 6점과 다른가, 그리고 6점보다 높은가?

#### 문제 설명

주택의 전반적인 품질을 나타내는 `OverallQual`의 기준 점수를 **6점**으로 설정했습니다.  
같은 표본으로 다음 두 연구 질문을 각각 검정하고 결과를 비교하세요.

- 질문 A: 평균 품질 점수가 6점과 **다른가?**
- 질문 B: 평균 품질 점수가 6점보다 **높은가?**

#### 요구사항

1. `OverallQual`에서 결측치를 제거하여 `overall_qual`에 저장하세요.
2. 표본 수와 표본평균을 출력하세요.
3. 질문 A의 가설을 작성하고 `alternative="two-sided"`로 양측검정을 수행하세요.
4. 질문 B의 가설을 작성하고 `alternative="greater"`로 단측검정을 수행하세요.

#### 해석 질문

**Q1.** 질문 A와 질문 B의 대립가설은 어떻게 다른가요?  
**Q2.** 양측검정과 단측검정의 결론은 각각 무엇인가요?  
**Q3.** 분석 결과를 먼저 확인한 뒤 p-value를 작게 만들기 위해 단측검정으로 바꿔도 되나요?  
**Q4.** 실제 평균 품질 점수가 6점인데도 검정 결과가 유의하다고 판단했다면 어떤 오류인가요?

#### 제출 결과

- 두 검정의 가설
- 표본 수와 표본평균
- 양측검정 및 단측검정 코드
- 두 검정의 t통계량과 p-value
- 검정별 귀무가설 판단
- Q1~Q4 답변

```python
# 필수 2 코드를 작성하세요.
# 1. 결측치 제거
overall_qual = df['OverallQual'].dropna()

# 2. 표본 수와 표본평균
n_sample = len(overall_qual)
mean_sample = overall_qual.mean()
print(f"표본 수: {n_sample}")
print(f"표본평균: {mean_sample:.4f}")

# 3. 질문 A 가설 (양측검정)
# H0: Ames 지역의 평균 OverallQual은 [기준값]이다.
# H1: Ames 지역의 평균 OverallQual은 [기준값]이 아니다.
alpha = 0.05
t_stat_A, p_value_A = stats.ttest_1samp(overall_qual, popmean=6, alternative='two-sided')

# 4. 질문 B 가설 (단측검정)
# H0: Ames 지역의 평균 OverallQual은 [기준값]이다. (또는 이하이다)
# H1: Ames 지역의 평균 OverallQual은 [기준값]보다 크다.
t_stat_B, p_value_B = stats.ttest_1samp(overall_qual, popmean=6, alternative='greater')

# 5. 결과 출력
print(f"[양측검정] t통계량: {t_stat_A:.4f}, p-value: {p_value_A:.4f}")
print(f"[단측검정] t통계량: {t_stat_B:.4f}, p-value: {p_value_B:.4f}")

# 6. 기각 여부
print("양측검정:", "기각" if p_value_A < alpha else "기각 실패")
print("단측검정:", "기각" if p_value_B < alpha else "기각 실패")
```

#### 해석 질문

**Q3.** 실제 평균이 1,550보다 작은데 귀무가설을 기각하지 못했다면 어떤 오류인가요?

#### 제출 결과

- 가설 설정과 검정 방향 선택 이유
- 표본 수와 표본평균

```python
# 과제 1 코드를 작성하세요.

# 1. GrLivArea 결측치 제거
gr_liv_area = df['GrLivArea'].dropna()

# 2. 표본 수와 표본평균 확인
n_sample = len(gr_liv_area)
mean_sample = gr_liv_area.mean()
print(f"표본 수: {n_sample}")
print(f"표본평균: {mean_sample:.4f}")

alpha = 0.05

# 3. 가설 설정
# H0: Ames 지역 주택의 평균 GrLivArea는 1550 이상이다.
# H1: Ames 지역 주택의 평균 GrLivArea는 1550보다 작다.

# 4. 검정 방향: 단측검정(alternative='less')
# 분석가의 주장이 "1550보다 작다"처럼 방향이 명확하므로 단측검정을 사용

# 5. 단일표본 t검정 수행
t_stat, p_value = stats.ttest_1samp(gr_liv_area, popmean=1550, alternative='less')

# 6. t통계량과 p-value 출력
print(f"t통계량: {t_stat:.4f}")
print(f"p-value: {p_value:.4f}")

# 7. 귀무가설 기각 여부 판단
if p_value < alpha:
    print(f"p-value({p_value:.4f}) < alpha({alpha}) → 귀무가설 기각")
else:
    print(f"p-value({p_value:.4f}) >= alpha({alpha}) → 귀무가설 기각 실패")
```

5. 제1종 오류와 제2종 오류는 각각 어떤 상황인가요?
-> 제1종 오류는 실제로 귀무가설이 참인데 이를 기각하는 오류
-> 제2종 오류는 실제로 귀무가설이 거짓인데 이를 기각하지 못하는 오류
