# 필수 2. 업로드한 데이터를 시계열 데이터로 정리하고 예측하기

## 배경

업로드한 Bike Sharing Demand 원본 데이터는 시간 단위 관측값을 포함하고 있습니다.

웹앱에서는 `datetime`을 날짜/시간으로 변환하고, 월 단위로 전체 대여량을 합계하여 시계열 분석용 데이터를 만듭니다.

## 문제 1. 월별 시계열 데이터를 만드세요.

### 요구사항

파일이 업로드된 경우 다음 작업을 수행합니다.

1. `datetime`을 `pd.to_datetime()`으로 변환합니다.
2. 날짜 순서대로 정렬합니다.
3. `datetime`에서 월 정보를 추출합니다.
4. 월별 `count` 합계를 계산합니다.
5. 월을 인덱스로 설정합니다.
6. `st.dataframe()`으로 월별 데이터를 표시합니다.
7. `st.line_chart()`로 월별 대여량 변화를 표시합니다.

### 확인 포인트

- 시간 순서가 유지되는가?
- 한 행이 한 달을 나타내는 데이터로 정리되었는가?
- 표와 선 그래프가 모두 표시되는가?

````python
# TODO
# 업로드한 원본 데이터를 월별 시계열 데이터로 정리하세요.

if uploaded_file is not None:
    data = pd.read_csv(uploaded_file)
    # 'date' 컬럼을 datetime 형식으로 변환
    data['datetime'] = pd.to_datetime(data['datetime'])
    data = data.sort_values('datetime')
    data['month'] = (data['datetime'].dt.to_period('M')).dt.to_timestamp()
    monthly = (data.groupby('month', as_index=False)['count'].sum())
    monthly = monthly.set_index('month')
    # 'date' 컬럼을 인덱스로 설정하고 월별로 그룹화하여 합계 계산
    st.write("월별 자전거 대여량")
    st.dataframe(monthly)
    st.write('시간에 따른 월별 대여량 변화')
    st.line_chart(monthly['count'])
````

# 과제 1. 시계열 데이터 예측 웹앱 완성하기

이번 과제에서는 앞에서 만든 기능을 하나의 `app.py`에 연결합니다.

> 새로운 기법을 추가하지 않고 필수 실습에서 사용한 기능을 연결하는 수준입니다.

## 요구사항

1. `"Bike Sharing 시계열 예측"` 제목을 표시합니다.
2. `Bike_Sharing_Demand.csv`를 업로드할 수 있도록 합니다.
3. 업로드한 데이터를 pandas DataFrame으로 불러옵니다.
4. `datetime`을 날짜/시간 자료형으로 변환합니다.
5. 시간 순서대로 정렬한 뒤 월별 `count` 합계를 계산합니다.
6. 월별 데이터를 표와 선 그래프로 표시합니다.
7. 사용자가 `1`, `3`, `6`개월 중 예측 기간을 선택합니다.
8. 전체 월별 `count`로 `ARIMA(2, 1, 1)`을 학습합니다.
9. 선택한 기간만큼 미래 값을 예측합니다.
10. 마지막 관측 월의 다음 달부터 미래 날짜를 생성합니다.
11. 미래 날짜와 예측값을 DataFrame으로 만들어 표로 표시합니다.
12. 기존 실제값과 미래 예측값을 하나의 선 그래프로 표시합니다.

### 확인 포인트

- 파일 업로드 → 데이터 처리 → 사용자 입력 → ARIMA 예측 → 결과 출력 흐름이 연결되는가?
- 미래 날짜의 개수가 사용자가 선택한 예측 기간과 같은가?
- 기존 실제값과 미래 예측값을 구분하여 표시하는가?
- 미래 실제값이 아직 없으므로 MAE와 RMSE를 계산하지 않았는가?

````python
# TODO
# 지금까지 작성한 코드를 하나의 app.py로 연결하세요.

import streamlit as st
import pandas as pd
from statsmodels.tsa.arima.model import ARIMA

st.title("Bike Sharing 시계열 예측")
st.write("월별 자전거 대여량을 확인하고 미래 값을 예측합니다.")

uploaded_file = st.file_uploader("CSV 파일을 업로드하세요.", type=["csv"])

if uploaded_file is not None:
    data = pd.read_csv(uploaded_file)
    st.write("업로드한 원본 데이터")
    st.dataframe(data)

    data["datetime"] = pd.to_datetime(data["datetime"])
    data = data.sort_values("datetime")
    data["month"] = data["datetime"].dt.to_period("M").dt.to_timestamp()

    monthly = data.groupby("month", as_index=False)["count"].sum()
    monthly = monthly.set_index("month")

    st.write("월별 자전거 대여량")
    st.dataframe(monthly)

    st.write("시간에 따른 월별 대여량 변화")
    st.line_chart(monthly["count"])

    forecast_period = st.selectbox("예측 기간을 선택하세요 (개월)", [1, 3, 6])

    model = ARIMA(monthly["count"], order=(2, 1, 1))
    fitted = model.fit()
    forecast = fitted.forecast(steps=forecast_period)

    last_month = monthly.index[-1]
    future_dates = pd.date_range(
        start=last_month + pd.DateOffset(months=1),
        periods=forecast_period,
        freq="MS",
    )

    forecast_df = pd.DataFrame(
        {"예측 대여량": forecast.values.round(0).astype(int)},
        index=future_dates,
    )
    forecast_df.index.name = "month"

    st.write(f"향후 {forecast_period}개월 예측 결과")
    st.dataframe(forecast_df)

    chart_df = pd.DataFrame({
        "실제 대여량": monthly["count"],
        "예측 대여량": forecast_df["예측 대여량"],
    })

    st.write("실제값과 예측값 비교")
    st.line_chart(chart_df)
````

### Q3. 이번 웹앱에서 MAE와 RMSE를 계산하지 않는 이유는 무엇인가요?

**답변:**  
- 예측 대상이 아직 관측되지 않은 미래이므로 비교할 실제값이 존재하지 않기 때문.  

MAE와 RMSE는 실제값과 예측값의 차이를 계산하는 지표이므로, 실제값이 있어야 계산할 수 있음.  

원래는 전체 데이터를 Train과 Test로 나누어 Test 구간의 실제값을 정답으로 두고 예측값과 비교했으나,  
웹앱은 전체 월별 데이터를 학습에 사용하고 마지막 관측 월 이후를 예측하므로 정답이 되는 실제값이 없음.  
-> 오차 지표 대신 과거 실제값과 예측값을 하나의 그래프로 표시하여 예측값이 기존 추세와 자연스럽게 이어지는지를 확인함.
