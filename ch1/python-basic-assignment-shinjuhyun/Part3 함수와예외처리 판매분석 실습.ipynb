# 문제 1. 판매금액 구간 분류 함수 만들기

## 함수

```python
classify_sales_by_amount(sale_ids, sale_amounts)
```

## 요구사항

1. 두 입력값이 리스트가 아니면 `TypeError`를 발생시킵니다.
2. 두 리스트의 길이가 다르면 `ValueError`를 발생시킵니다.
3. 데이터가 비어 있으면 `ValueError`를 발생시킵니다.
4. 고가·중가·일반 거래의 `SaleID`를 딕셔너리로 반환합니다.
5. 함수를 호출하고 결과를 출력합니다.

````python
# 문제 1 코드를 작성하세요.

def classify_sales_by_amount(sale_ids, sale_amounts):
    if type(sale_ids) != list or type(sale_amounts) != list:
        raise TypeError ("TypeError : 두 입력값 리스트 X")
    elif len(sale_ids) != len(sale_amounts):
        raise ValueError ("ValueError : 두 리스트 길이가 다름")
    elif len(sale_ids) == 0:
        raise ValueError ("ValueError : 데이터 비어있음")

    low_id = []
    middle_id = []
    high_id = []

    for count, amount in enumerate(sale_amounts):
        Sale_ID = sale_ids[count]
        if amount >= 70000000:
            high_id.append(Sale_ID)
        elif amount >= 40000000 and amount < 70000000:
            middle_id.append(Sale_ID)
        else:
            low_id.append(Sale_ID)
    return {'고가':high_id, '중가': middle_id, '일반': low_id}

result = classify_sales_by_amount(sale_ids, sale_amounts)

print(f"고가 거래 수 : {len(result['고가'])}, sale_id : {result['고가']}")
print(f"중가 거래 수 : {len(result['중가'])}, sale_id : {result['중가']}")
print(f"일반 거래 수 : {len(result['일반'])}, sale_id : {result['일반']}")   
````

# 문제 2. 배송 확인 거래 검색 함수 만들기

## 함수

```python
find_delivery_check_sales(
    sale_ids,
    delivery_days,
    stock_statuses
)
```

## 요구사항

1. 세 입력값이 리스트가 아니면 `TypeError`를 발생시킵니다.
2. 세 리스트의 길이가 다르면 `ValueError`를 발생시킵니다.
3. 배송일 20일 이상이면서 출고완료가 아닌 거래의 `SaleID`를 반환합니다.
4. 정상 입력과 잘못된 입력을 각각 테스트합니다.
5. `TypeError`와 `ValueError`를 별도의 `except`에서 처리합니다.

````python
# 문제 2 코드를 작성하세요.
def find_delivery_check_sales(sale_ids, delivery_days, stock_statuses):
        if type(sale_ids) != list or type(delivery_days) != list or type(stock_statuses) != list:
            raise TypeError ("리스트가 아닙니당")
        elif len(sale_ids) != len(delivery_days) or len(sale_ids) != len(stock_statuses):
            raise ValueError ("리스트 길이가 다ㅡ릅니다")
        #아이디 구하기
        SaleID = []
        for count, day in enumerate(delivery_days):
            if day >= 20 and stock_statuses[count] != "출고완료":
                 SaleID.append(sale_ids[count])
        return SaleID

try:
    result = find_delivery_check_sales(
        ['S1', 'S2', 'S3', 'S4'],
        [25, 10, 22, 30],
        ['배송중', '출고완료', '배송중', '재고부족']
    )
    print("테스트1 (정상 입력) 결과:", result)
except TypeError as e:
    print("테스트1 TypeError:", e)
except ValueError as e:
    print("테스트1 ValueError:", e)

try:
    result = find_delivery_check_sales(
        ('S1', 'S2', 'S3'),
        [25, 10, 22],
        ['배송중', '출고완료', '배송중']
    )
    print("테스트2 결과:", result)
except TypeError as e:
    print("테스트2 TypeError:", e)
except ValueError as e:
    print("테스트2 ValueError:", e)
````
