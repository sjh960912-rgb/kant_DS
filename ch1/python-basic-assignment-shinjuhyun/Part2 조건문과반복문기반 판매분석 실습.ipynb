## 제공 코드. 판매 데이터 불러오기

아래 코드는 준비 코드이며 평가 대상이 아닙니다.

```python
import pandas as pd

DATA_FILE = "자동차_판매_데이터.csv"

df = pd.read_csv(DATA_FILE)

sale_ids = df["SaleID"].tolist()
sale_amounts = df["FinalSaleAmount"].tolist()
delivery_days = df["DeliveryDays"].tolist()
stock_statuses = df["StockStatus"].tolist()
returned_flags = df["IsReturned"].tolist()

print("전체 거래 수:", len(df))
df.head()
```

# 문제 1. 판매금액 구간별 거래 분류

## 업무 상황

영업관리팀은 고가·중가·일반 거래 규모를 확인하여 현재 판매 현황을 점검하려고 합니다.

## 분류 기준

| 거래 구간 | 판매금액 기준 |
|---|---:|
| 고가 거래 | 70,000,000원 이상 |
| 중가 거래 | 40,000,000원 이상 70,000,000원 미만 |
| 일반 거래 | 40,000,000원 미만 |

## 요구사항

1. `sale_amounts`를 반복합니다.
2. `if-elif-else`로 판매금액을 하나의 구간에 분류합니다.
3. 각 구간의 거래 수를 변수에 저장합니다.
4. 각 구간의 `SaleID`를 별도 리스트에 저장합니다.
5. 구간별 거래 수와 `SaleID` 앞 5개를 출력합니다.
6. 세 구간의 거래 수 합계가 전체 거래 수와 같은지 확인합니다.

```python
# 문제 1 코드를 작성하세요.
#필요한 것 : 구간별 거래 수, sale id

low_count = 0
middle_count = 0
high_count = 0

low_id = []
middle_id = []
high_id = []


for count, amount in enumerate(sale_amounts):
    Sale_ID = sale_ids[count]
    if amount >= 70000000:
        high_count += 1
        high_id.append(Sale_ID)
    elif amount >= 40000000 and amount < 70000000:
        middle_count +=1
        middle_id.append(Sale_ID)
    else:
        low_count += 1
        low_id.append(Sale_ID)

total = low_count + middle_count + high_count

print(f"고가 구간별 거래 수: {high_count}, Sale_ID: {high_id[:5]}")
print(f"중가 구간별 거래 수: {middle_count}, Sale_ID: {middle_id[:5]}")
print(f"일반 구간별 거래 수: {low_count}, Sale_ID: {low_id[:5]}")

print("세 구간 거래 수:", total)
print("전체 거래 수:", len(df))
```

# 문제 2. 배송 확인이 필요한 거래 찾기

## 업무 상황

배송 기간이 길지만 아직 출고되지 않은 거래를 영업관리팀이 우선 확인해야 합니다.

## 조건

- `DeliveryDays`가 20일 이상
- `StockStatus`가 `"출고완료"`가 아님

## 요구사항

1. `range(len(df))`를 사용해 거래 위치를 반복합니다.
2. 같은 위치의 배송일과 출고 상태를 확인합니다.
3. 두 조건을 만족하면 거래 수를 1 증가시킵니다.
4. 해당 거래의 `SaleID`를 리스트에 저장합니다.
5. 거래 수와 `SaleID` 앞 5개를 출력합니다.

```python
# 문제 2 코드를 작성하세요.
#필요한 것 : 거래 수 , sale id

count = 0
id = []

for loop in range(len(df)):
    day = delivery_days[loop]
    Sale_ID = sale_ids[loop]
    status = stock_statuses[loop]
    if day >= 20:
        if status != "출고완료":
            count += 1
            id.append(Sale_ID)

print("배송 확인 거래 수:", count)
print("Sale ID:", id[:5])


```

# 문제 3. 고액 반품 거래 찾기

## 업무 상황

판매금액이 큰 반품 거래는 환불 금액과 고객 대응 영향이 크므로 우선 점검 대상입니다.

## 조건

- `FinalSaleAmount`가 70,000,000원 이상
- `IsReturned`가 `"Y"`

## 요구사항

1. 전체 거래 위치를 반복합니다.
2. 판매금액과 반품 여부를 함께 확인합니다.
3. 두 조건을 만족하면 거래 수를 1 증가시킵니다.
4. 해당 거래의 `SaleID`를 리스트에 저장합니다.
5. 거래 수와 전체 `SaleID`를 출력합니다.

```python
# 문제 3 코드를 작성하세요.
#필요한 것 판매 금액, 반품 여부 / 거래수, id

refund_count = 0
refund_id = []

for loop in range(len(df)):
    amount = sale_amounts[loop]
    flag = returned_flags[loop]
    id = sale_ids[loop]
    if amount >= 70000000 and flag == "Y":
        refund_count += 1
        refund_id.append(id)

print("고액 반품 거래 수:", refund_count)
print("Sale ID:", refund_id)


```
