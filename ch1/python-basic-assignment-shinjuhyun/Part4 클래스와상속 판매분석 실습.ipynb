# 문제 1. 판매 거래를 표현하는 `CarSale` 클래스 만들기

## 업무 상황

Part 1과 Part 2에서는 판매 데이터를 여러 리스트와 함수로 관리했습니다.  
이번에는 판매 거래 한 건의 정보와 점검 기능을 하나의 객체로 관리하기 위해 `CarSale` 클래스를 작성합니다.

## 요구사항

1. `CarSale` 클래스를 정의합니다.
2. `__init__()` 생성자를 작성하고 다음 값을 매개변수로 받습니다.
   - `sale_id`
   - `manufacturer`
   - `model`
   - `final_sale_amount`
   - `delivery_days`
   - `stock_status`
   - `is_returned`
3. 생성자로 전달받은 값을 같은 이름의 인스턴스 속성에 저장합니다.
4. `get_amount_category()` 메서드를 작성합니다.
   - 판매금액이 70,000,000원 이상이면 `"고가"`를 반환합니다.
   - 판매금액이 40,000,000원 이상 70,000,000원 미만이면 `"중가"`를 반환합니다.
   - 판매금액이 40,000,000원 미만이면 `"일반"`을 반환합니다.
5. `needs_delivery_check()` 메서드를 작성합니다.
   - 배송일이 20일 이상이고 출고 상태가 `"출고완료"`가 아니면 `True`를 반환합니다.
   - 그 외에는 `False`를 반환합니다.
6. `is_high_value_return(min_amount=70_000_000)` 메서드를 작성합니다.
   - 판매금액이 `min_amount` 이상이고 반품 여부가 `"Y"`이면 `True`를 반환합니다.
   - 그 외에는 `False`를 반환합니다.
7. 다음 값을 사용하여 `CarSale` 객체 한 개를 생성합니다.
   - 거래 ID: `TEST001`
   - 제조사: `현대`
   - 모델: `테스트차량`
   - 판매금액: 75,000,000원
   - 배송일: 25일
   - 출고 상태: `"재고있음"`
   - 반품 여부: `"Y"`
8. 생성한 객체의 `sale_id`, `manufacturer`, `model` 속성을 출력합니다.
9. 세 메서드를 각각 호출하고 반환 결과를 출력합니다.

## ✅ 확인 포인트

- 생성자에서 전달받은 값이 객체 속성에 정확히 저장되어야 합니다.
- 판매금액 기준에 따라 거래 구간이 올바르게 반환되어야 합니다.
- 배송 점검과 고액 반품 조건에 따라 `True` 또는 `False`가 반환되어야 합니다.
- 생성한 객체에서 속성과 메서드를 정상적으로 호출할 수 있어야 합니다.


````python
# 문제 1 코드를 작성하세요.

class CarSale:
    def __init__(self, sale_id, manufacturer, model, final_sale_amount, delivery_days, stock_status, is_returned):
        self.sale_id = sale_id
        self.manufacturer = manufacturer
        self.model = model
        self.final_sale_amount = final_sale_amount
        self.delivery_days = delivery_days
        self.stock_status = stock_status
        self.is_returned = is_returned

    def get_amount_category(self):
        if self.final_sale_amount >= 70000000:
            return "고가"
        elif self.final_sale_amount >= 40000000:
            return "중가"
        elif self.final_sale_amount < 40000000:
            return "일반"
            
    def needs_delivery_check(self):
        if self.delivery_days >= 20 and self.stock_status != "출고완료":
            return True
        else:
            return False

    def is_high_value_return(self,min_amount=70_000_000):
        min_amount = 70000000
        if self.final_sale_amount >= min_amount and self.is_returned == "Y":
            return True
        else:
            return False
result = CarSale(sale_id= "TEST001", manufacturer= "현대", model= "테스트차량", final_sale_amount= 75000000, 
           delivery_days= 25, stock_status= "재고보유", is_returned= "Y")



#result = CarSale(["TEST001"],["현대"], ["테스트차량"], ["75000000"], ["25"], ["재고있음"], ["Y"])

print(type(result.sale_id))
print(type(result.manufacturer))
print(type(result.model))

print(result.sale_id)
print(result.manufacturer)
print(result.model)

print("판매 금액: ",result.get_amount_category())
print("배송 점검: ",result.needs_delivery_check())
print("고액 반품: ",result.is_high_value_return())
````

# 문제 2. 여러 판매 객체를 관리하는 `CarSalesManager` 클래스 만들기

## 업무 상황

판매 거래가 여러 건으로 늘어나면 개별 객체를 하나씩 확인하기 어렵습니다.  
여러 `CarSale` 객체를 하나의 목록으로 관리하고, 거래 현황을 요약할 수 있는 `CarSalesManager` 클래스를 작성합니다.

## 요구사항

1. `CarSalesManager` 클래스를 정의합니다.
2. `__init__(self, sales)` 생성자를 작성합니다.
3. 전달받은 `CarSale` 객체 리스트를 `self.sales` 속성에 저장합니다.
4. `count_by_amount_category()` 메서드를 작성합니다.
   - 모든 판매 객체의 `get_amount_category()`를 호출합니다.
   - 고가·중가·일반 거래 수를 각각 계산합니다.
   - 계산 결과를 딕셔너리로 반환합니다.
5. `find_delivery_check_sales()` 메서드를 작성합니다.
   - 각 판매 객체의 `needs_delivery_check()`를 호출합니다.
   - 결과가 `True`인 객체만 리스트에 담아 반환합니다.
6. `find_high_value_returns()` 메서드를 작성합니다.
   - 각 판매 객체의 `is_high_value_return()`을 호출합니다.
   - 결과가 `True`인 객체만 리스트에 담아 반환합니다.
7. `get_summary()` 메서드를 작성합니다.
   - 전체 거래 수를 `{"total_sales": 거래 수}` 형태의 딕셔너리로 반환합니다.
8. DataFrame의 각 행을 반복하여 `CarSale` 객체로 변환합니다.
9. 생성한 모든 `CarSale` 객체를 리스트에 저장합니다.
10. 객체 리스트를 사용하여 `CarSalesManager` 객체를 생성합니다.
11. 다음 결과를 출력합니다.
    - 전체 거래 수
    - 판매금액 구간별 거래 수
    - 배송 확인 대상 거래 수와 `SaleID` 앞 5개
    - 고액 반품 거래 수와 전체 `SaleID`

## ✅ 확인 포인트

- DataFrame의 각 행이 하나의 `CarSale` 객체로 변환되어야 합니다.
- 여러 판매 객체가 `CarSalesManager`의 `sales` 속성에 저장되어야 합니다.
- 구간별 거래 수와 점검 대상 거래가 정확히 계산되어야 합니다.
- 각 메서드는 요구된 형식의 딕셔너리 또는 객체 리스트를 반환해야 합니다.


````python
# 문제 2 코드를 작성하세요.

class CarSalesManager:
    def __init__(self, sales):
        self.sales = sales

    def count_by_amount_category(self): #4
        count = {"고가" : 0, "중가" : 0, "일반" : 0}

        for sale in self.sales:
            count[sale.get_amount_category()] +=1
        return count
    
    def find_delivery_check_sales(self): #5
        delivery_result = []
        for sale in self.sales:
            if sale.needs_delivery_check() == True:
                delivery_result.append(sale.sale_id)
        return delivery_result

    def find_high_value_returns(self): #6
        high_value_result = []
        for sale in self.sales:
            if sale.is_high_value_return() == True:
                high_value_result.append(sale.sale_id)
        return high_value_result

    def get_summary(self): #7
        return{"total_sales": len(self.sales)}

sales = []

for index, row in df.iterrows(): #7
    sale = CarSale(
        sale_id=row["SaleID"],
        manufacturer=row["Manufacturer"],
        model=row["Model"],
        final_sale_amount=row["FinalSaleAmount"],
        delivery_days=row["DeliveryDays"],
        stock_status=row["StockStatus"],
        is_returned=row["IsReturned"]
    )

    sales.append(sale)

manager = CarSalesManager(sales)

print("전체 거래 수:", manager.get_summary()["total_sales"])
print("판매금액 구간별 거래 수:", manager.count_by_amount_category())

delivery_sales = manager.find_delivery_check_sales()

print("배송 확인 대상 거래 수:", len(delivery_sales))
print("배송 확인 대상 SaleID 앞 5개:", delivery_sales[:5])

high_value_returns = manager.find_high_value_returns()

print("고액 반품 거래 수:", len(high_value_returns))
print( "고액 반품 거래 전체 SaleID:", high_value_returns)
````

# 문제 3. 상속을 활용한 `PremiumCarSalesManager` 클래스 만들기

## 업무 상황

영업관리팀은 전체 거래 관리 기능과 별도로 고액 거래만 집중적으로 관리하려고 합니다.  
기존 `CarSalesManager`의 기능을 다시 작성하지 않고 상속하여 고액 거래 전용 기능을 추가합니다.

## 요구사항

1. `CarSalesManager`를 상속하는 `PremiumCarSalesManager` 클래스를 정의합니다.
2. `__init__(self, sales, premium_threshold=70_000_000)` 생성자를 작성합니다.
3. `super().__init__(sales)`를 호출하여 부모 클래스의 판매 객체 저장 기능을 재사용합니다.
4. 고액 거래 기준값을 `self.premium_threshold` 속성에 저장합니다.
5. `find_premium_sales()` 메서드를 작성합니다.
   - 모든 판매 객체를 반복합니다.
   - `final_sale_amount`가 `premium_threshold` 이상인 객체만 리스트에 저장합니다.
   - 고액 거래 객체 리스트를 반환합니다.
6. `get_premium_summary()` 메서드를 작성합니다.
   - `find_premium_sales()`를 호출하여 고액 거래를 찾습니다.
   - 고액 거래 수를 계산합니다.
   - 고액 거래의 총 판매금액을 계산합니다.
   - 다음 형식의 딕셔너리로 반환합니다.

```python
{
    "premium_count": 고액_거래_수,
    "premium_total_amount": 고액_거래_총판매금액
}
```

7. 문제 2에서 만든 판매 객체 리스트를 사용하여 `PremiumCarSalesManager` 객체를 생성합니다.
8. 고액 거래 기준, 고액 거래 수, 고액 거래 총 판매금액을 출력합니다.
9. 고액 거래의 `SaleID` 앞 5개를 출력합니다.

## ✅ 확인 포인트

- 자식 클래스가 `CarSalesManager`를 정확히 상속해야 합니다.
- `super().__init__()`를 사용하여 부모 생성자를 호출해야 합니다.
- 고액 거래 기준값이 자식 객체의 속성으로 저장되어야 합니다.
- 기준금액 이상인 거래만 고액 거래로 반환되어야 합니다.
- 고액 거래 수와 총 판매금액이 정확히 계산되어야 합니다.


````python
# 문제 3 코드를 작성하세요.
class PremiumCarSalesManager(CarSalesManager):
    def __init__(self, sales, premium_threshold=70_000_000):
        super().__init__(sales)
        self.premium_threshold = premium_threshold

    def find_premium_sales(self):
        premium_sales = []

        for sale in self.sales:
            if sale.final_sale_amount >= self.premium_threshold:
                premium_sales.append(sale)
        return premium_sales
    
    def get_premium_summary(self):
        premium_sales = self.find_premium_sales()
        count = len(premium_sales)
        total_amount = 0

        for sale in premium_sales:
            total_amount += sale.final_sale_amount
        return {"premium_sales_count" : count, "premium_sales_total_amount" : total_amount}

premium_manager = PremiumCarSalesManager(sales)
premium_summary = premium_manager.get_premium_summary()

print("고액 거래 기준:", premium_manager.premium_threshold)
print("고액 거래 수:", premium_summary["premium_sales_count"])
print("고액 거래 총 판매금액:", premium_summary["premium_sales_total_amount"])


# 9. 고액 거래 SaleID 앞 5개 출력
premium_sales = premium_manager.find_premium_sales()

print("고액 거래 SaleID 앞 5개:", [sale.sale_id for sale in premium_sales[:5]])
````
